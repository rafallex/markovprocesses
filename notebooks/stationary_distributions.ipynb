```python
import numpy as np

def null_space(A, tol=1e-12):
    """Compute an orthonormal basis for the null space of A using SVD."""
    U, S, Vt = np.linalg.svd(A)
    rank = (S > tol).sum()
    ns = Vt[rank:].T
    return ns

def stationary_distributions(P):
    """Return normalized null-space vectors for (P - I)^T."""
    n = P.shape[0]
    A = (P - np.eye(n)).T
    v = null_space(A)
    # Normalize each basis vector to sum to 1 (when possible)
    pis = []
    for k in range(v.shape[1]):
        vec = v[:, k]
        s = vec.sum()
        if abs(s) < 1e-12:
            # Skip degenerate case; user can inspect v directly
            continue
        pi = vec / s
        pis.append(pi)
    return v, pis
```

## Example 1
$$P = \begin{pmatrix}0.6 & 0.4\\ 0.2 & 0.8\end{pmatrix}$$

```python
P = np.array([[0.6, 0.4],
              [0.2, 0.8]])

v, pis = stationary_distributions(P)
print("Null-space basis (columns):")
print(v)
print("\nStationary distribution(s):")
for i, pi in enumerate(pis, 1):
    print(f"pi{i} = {pi}")

# Verify pi P = pi
for i, pi in enumerate(pis, 1):
    print(f"pi{i} P = {pi @ P}")
```

## Example 2
$$P = \begin{pmatrix}0.6 & 0.4 & 0\\ 0.2 & 0.8 & 0\\ 0 & 0 & 1\end{pmatrix}$$

```python
P = np.array([[0.6, 0.4, 0.0],
              [0.2, 0.8, 0.0],
              [0.0, 0.0, 1.0]])

v, pis = stationary_distributions(P)
print("Null-space basis (columns):")
print(v)
print("\nStationary distribution(s):")
for i, pi in enumerate(pis, 1):
    print(f"pi{i} = {pi}")

# Verify pi P = pi
for i, pi in enumerate(pis, 1):
    print(f"pi{i} P = {pi @ P}")
```

## General method (matches MATLAB vectorization)
This mirrors the MATLAB snippet:
- `v = null((P - I)^T)`
- `pi = v .* (ones(n,1) * (1./sum(v)))`

```python
def stationary_matrix(P):
    """Return matrix whose columns are normalized stationary vectors."""
    n = P.shape[0]
    v = null_space((P - np.eye(n)).T)
    if v.size == 0:
        return v
    col_sums = v.sum(axis=0)
    # Broadcast to normalize columns (avoid division by zero)
    col_sums = np.where(np.abs(col_sums) < 1e-12, np.nan, col_sums)
    pi = v / col_sums
    return pi

P = np.array([[0.6, 0.4, 0.0],
              [0.2, 0.8, 0.0],
              [0.0, 0.0, 1.0]])
pi = stationary_matrix(P)
print("Stationary matrix (columns):")
print(pi)
```
